```python
# ============================================================
# TEXT PREPROCESSING IN PYTHON (GOOGLE COLAB VERSION)
# Domain: Public Finance in Kenya (my domain area)
# ============================================================

# -------------------------------
# SECTION 0: Import Required Libraries for NLP
# -------------------------------

import nltk
import re
import string

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download required NLTK resources
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# -------------------------------
# SECTION 1: Original Text (Noisy Data with all manner of "impurities")
# -------------------------------

# Sentence from public finance domain in Kenya
# Contains emojis, numbers, punctuation, capital letters, etc.
original_text = """
In 2025/2026, Kenya’s National Assembly allocated KSh 1.9 Billion 💰 to
the Ministry of Health for health-sector reforms!!! However, 25% of the funds
were delayed due to revenue underperformance 😕 #Budget2025
"""

print("===== ORIGINAL TEXT =====")
print(original_text)

# -------------------------------
# SECTION 2: Convert Text to Lowercase
# -------------------------------

lower_text = original_text.lower()

print("\n===== AFTER LOWERCASE =====")
print(lower_text)

## -------------------------------
# SECTION 3: Remove Numbers, Punctuation, and Emojis
# -------------------------------

# Remove numbers
text_no_numbers = re.sub(r'\d+', '', lower_text)

# Remove punctuation
text_no_punct = text_no_numbers.translate(
    str.maketrans('', '', string.punctuation)
)

# Remove emojis and other non-text Unicode symbols
# This regex keeps only letters and spaces
text_no_emoji = re.sub(r'[^a-zA-Z\s]', '', text_no_punct)

# Remove extra whitespace
text_clean_basic = re.sub(r'\s+', ' ', text_no_emoji).strip()

print("\n===== AFTER REMOVING NUMBERS, PUNCTUATION & EMOJIS =====")
print(text_clean_basic)

# -------------------------------
# SECTION 4: Remove Stopwords
# -------------------------------

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Tokenize text (split into words)
tokens = text_clean_basic.split()

# Remove stopwords
filtered_tokens = [word for word in tokens if word not in stop_words]

print("\n===== AFTER STOPWORD REMOVAL =====")
print(" ".join(filtered_tokens))

# -------------------------------
# SECTION 5: Apply Lemmatization (I use WordNetLemmatizer)
# -------------------------------

lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]

cleaned_text = " ".join(lemmatized_tokens)

print("\n===== AFTER LEMMATIZATION =====")
print(cleaned_text)

# -------------------------------
# SECTION 6: Display Original vs Cleaned Text
# -------------------------------

print("\n================================================")
print("FINAL COMPARISON")
print("================================================")

print("\nOriginal Text:")
print(original_text)

print("\nCleaned Text:")
print(cleaned_text)
```
